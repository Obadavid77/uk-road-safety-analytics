## Summary of Key Findings

### Finding 1 — Serious Casualties Have Not Come Down Since COVID
Serious casualties rose from 24,493 in 2020 to 31,942 in 2022 and 
have stayed at that elevated level through 2024 (32,071). Fatal 
casualties peaked at 2,656 in 2022 and remain elevated at 2,508 in 
2024. The government's target to halve killed and seriously injured 
casualties by 2030 looks very difficult to achieve without significant 
new intervention.

### Finding 2 — 5pm on a Friday is the Most Dangerous Time on British Roads
The evening rush hour (4-7pm) consistently produces the most casualties 
across all 5 years. 5pm alone accounts for 55,216 casualties over the 
period. Friday is the most dangerous day with 105,883 casualties — 39% 
more than Sunday. Overnight accidents (midnight to 5am) are fewer in 
number but disproportionately fatal or serious.

### Finding 3 — 60mph Roads Kill More Than 70mph Roads
Roads with a 60mph speed limit have a 37.3% fatal and serious rate — 
higher than 70mph motorways (27.4%). Rural single carriageways at 60mph 
are the deadliest road type in Great Britain because two lanes of fast 
traffic share the same surface with no central barrier. Flooding is the 
most dangerous surface condition at 30.4% fatal and serious rate.

### Finding 4 — Summer Has More Accidents but Higher Severity
August has the highest fatal and serious rate at 27.1% despite not 
being the busiest month for total casualties. October has the most 
total casualties (58,143). Winter months (January, February) have 
fewer accidents and lower severity rates — drivers slow down in cold 
conditions but not in dry summer sunshine.

## Business Recommendations

### Recommendation 1 — Reduce Speed Limits on Rural Single Carriageways
The data is unambiguous. 60mph rural single carriageways produce the 
highest fatal and serious casualty rates in Great Britain — worse than 
motorways travelling at 70mph. The Department for Transport should 
conduct a systematic review of rural A-road speed limits with a view 
to reducing the default from 60mph to 50mph on roads without central 
barriers. The evidence from this analysis suggests this single change 
would have the greatest impact on fatal casualty rates of any 
infrastructure intervention available.

### Recommendation 2 — Target Enforcement at Friday Evening Rush Hour
With 5pm Friday consistently the peak danger window across all 5 years 
of data, police forces and road safety campaigns should concentrate 
resources at this specific time. This is not guesswork — it is a 
reproducible, data-confirmed pattern. Targeted speed enforcement, 
increased police visibility, and drink-drive campaigns aimed at Friday 
evening would reach the highest-risk window on British roads.

### Recommendation 3 — Improve Emergency Response Times in Rural Areas
Rural location is the third strongest predictor of fatal and serious 
outcomes. One major reason is emergency response time — ambulances take 
longer to reach rural accident scenes, and time to treatment is critical 
for survival in serious collisions. NHS ambulance trusts and the 
Department for Transport should jointly analyse accident locations 
against ambulance response time data to identify rural hotspots where 
faster response could save lives.

### Recommendation 4 — Reframe Summer Road Safety Campaigns
Current road safety campaigns concentrate heavily on winter driving — 
ice, fog, and dark mornings. The data shows August has the highest 
fatal and serious casualty rate of any month. Summer campaigns should 
target rural speeding, motorcycle safety (more riders on the road in 
warm weather), and the false confidence that dry sunny conditions 
create. The message should be: dry roads are not safe roads.

### Recommendation 5 — Apply for Access to Sensitive STATS19 Data
The model's performance ceiling (ROC-AUC 0.634) is determined by the 
absence of contributory factor data from the public dataset. DfT 
publishes sensitive fields — including alcohol involvement, driver 
distraction, and road defects — separately, accessible by application. 
A follow-up analysis using this data could significantly improve 
predictive accuracy and identify the behavioural factors behind the 
structural patterns this analysis has revealed.

## Limitations and Further Research

### Limitations of This Analysis

- **Public data only** — the most informative STATS19 fields (contributory 
  factors, alcohol involvement, driver distraction) are held separately 
  and require a formal application to DfT. Their absence explains the 
  model performance ceiling at ROC-AUC 0.634.

- **Reported accidents only** — STATS19 only captures accidents reported 
  to police. Minor accidents are significantly under-reported. The dataset 
  therefore over-represents serious and fatal casualties relative to the 
  true total of all road accidents.

- **Great Britain only** — Northern Ireland uses a separate reporting 
  system and is not included in STATS19 data.

- **COVID distortion in 2020** — traffic volumes dropped dramatically 
  during lockdowns, making 2020 figures unrepresentative of normal road 
  use. Comparisons using 2020 as a baseline should be treated with caution.

- **Severity reporting changes** — some police forces switched from 
  self-reported to injury-based severity assessment during this period, 
  which affects the comparability of serious casualty figures across years. 
  DfT publishes adjusted figures to account for this but the adjustment 
  introduces its own uncertainty.

- **No driver or vehicle data** — age, experience, vehicle type, vehicle 
  age, and whether seatbelts were worn are not in the public dataset. 
  These are known to be significant predictors of severity.

### Suggested Further Research

- Apply for DfT sensitive data access to include contributory factors 
  and alcohol involvement in the model
- Link accident locations to road infrastructure data (road condition 
  ratings, lighting provision) to test whether infrastructure quality 
  predicts severity independently of speed limit
- Build a Streamlit web application that allows road safety teams to 
  input accident conditions and receive a real-time severity risk score
- Extend analysis to include cyclist and pedestrian casualties separately, 
  since their risk factors differ substantially from vehicle occupants
