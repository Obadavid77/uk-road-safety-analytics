## Summary of Key Findings

### Finding 1 — Serious Casualties Have Not Come Down Since COVID
Serious casualties rose from 24,493 in 2020 to 31,942 in 2022 and 
have stayed at that elevated level through 2024 (32,071). Fatal 
casualties peaked at 2,656 in 2022 and remain elevated at 2,508 in 
2024. The government's target to halve killed and seriously injured 
casualties by 2030 looks very difficult to achieve without significant 
new intervention.

### Finding 2 — 5pm on a Friday is the Most Dangerous Time on British Roads
The evening rush hour (4-7pm) consistently produces the most casualties 
across all 5 years. 5pm alone accounts for 55,216 casualties over the 
period. Friday is the most dangerous day with 105,883 casualties — 39% 
more than Sunday. Overnight accidents (midnight to 5am) are fewer in 
number but disproportionately fatal or serious.

### Finding 3 — 60mph Roads Kill More Than 70mph Roads
Roads with a 60mph speed limit have a 37.3% fatal and serious rate — 
higher than 70mph motorways (27.4%). Rural single carriageways at 60mph 
are the deadliest road type in Great Britain because two lanes of fast 
traffic share the same surface with no central barrier. Flooding is the 
most dangerous surface condition at 30.4% fatal and serious rate.

### Finding 4 — Summer Has More Accidents but Higher Severity
August has the highest fatal and serious rate at 27.1% despite not 
being the busiest month for total casualties. October has the most 
total casualties (58,143). Winter months (January, February) have 
fewer accidents and lower severity rates — drivers slow down in cold 
conditions but not in dry summer sunshine.
